### Training a Simple Machine Learning Model

Let's demonstrate a basic machine learning workflow using `scikit-learn`.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.datasets import make_classification

# 1. Generate some synthetic data for a classification task
X, y = make_classification(n_samples=1000, n_features=20, n_informative=10, n_redundant=5, random_state=42)

print(f"Shape of features (X): {X.shape}")
print(f"Shape of target (y): {y.shape}")
print(f"First 5 target values: {y[:5]}")
```

We've generated a synthetic dataset with 1000 samples and 20 features, where 10 features are informative for predicting the target variable `y`.

```python
# 2. Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)l

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Test set size: {X_test.shape[0]} samples")
```

The data has been split into a training set (70%) and a testing set (30%). The training set will be used to teach the model, and the test set will be used to evaluate its performance on unseen data.

```python
# 3. Choose and instantiate a simple classification model (Logistic Regression)
model = LogisticRegression(random_state=42, solver='liblinear')

# 4. Train the model using the training data
print("Training the model...")
model.fit(X_train, y_train)
print("Model training complete.")
```

Here, we're using Logistic Regression, a common algorithm for binary classification, and training it on our prepared training data.

```python

y_pred = model.predict(X_test)


accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"\nModel Accuracy: {accuracy:.4f}")
print("\nClassification Report:\n", report)
```

```python
from langchain_community.document_loaders import TextLoader

# Instantiate the TextLoader with the path to 'cricket.txt'
loader = TextLoader('cricket.txt')

# Load the documents
documents = loader.load()

# Print the content of the loaded document
print("Loaded document content:")
#for doc in documents:
    #print(doc.page_content)


print("here is the details of documents of doc file ", documents)

print("one paragraph gap of paging")

#print(doc.page_content)
print(type(documents))


print(len(documents))
print(documents[0])




print(documents[0].page_content , documents[0].metadata)


```

```python
from langchain_text_splitters import CharacterTextSplitter

# Initialize the CharacterTextSplitter
# You can adjust chunk_size and chunk_overlap as needed
text_splitter = CharacterTextSplitter(
    separator="\n\n", # Split by double newline to separate paragraphs
    chunk_size=200,    # Each chunk will have a maximum of 200 characters
    chunk_overlap=20,  # Overlap between chunks to maintain context
    length_function=len,
    is_separator_regex=False,
)

# Split the loaded documents into smaller chunks
chunks = text_splitter.split_documents(documents)

print(f"Number of original documents: {len(documents)}")
print(f"Number of chunks after splitting: {len(chunks)}")

print("\nFirst 3 chunks:")
for i, chunk in enumerate(chunks[:3]):
    print(f"--- Chunk {i+1} ---")
    print(chunk.page_content)
    print(f"Metadata: {chunk.metadata}")
```

```python
from fpdf import FPDF

def create_pdf_from_text(filename, text_content):
    pdf = FPDF()
    pdf.add_page()
    pdf.set_font('Arial', 'B', 16)
    pdf.cell(40, 10, 'Sample PDF Document')
    pdf.ln(10)
    pdf.set_font('Arial', '', 12)
    pdf.multi_cell(0, 10, text_content)
    pdf.output(filename)
    print(f"'{filename}' created successfully.")

# Content for our dummy PDF
dummy_text = """This is a sample PDF document created for demonstration purposes. It contains some text that will be loaded by PyPDFLoader. PyPDFLoader is part of Langchain's document loaders and is used to extract text content from PDF files. It's a handy tool for processing unstructured data stored in PDFs for various NLP tasks. We can see how it handles multiple paragraphs and basic formatting. This allows us to integrate PDF documents into our Langchain-based applications, such as for question-answering or summarization. The process involves initializing the loader, calling the load() method, and then accessing the page_content attribute of the resulting documents."""

# Create the PDF file
pdf_filename = 'sample.pdf'
create_pdf_from_text(pdf_filename, dummy_text)
```

```python
from langchain_community.document_loaders import PyPDFLoader

# Instantiate PyPDFLoader with the path to the PDF file
pdf_loader = PyPDFLoader(pdf_filename)

# Load the documents from the PDF
pdf_documents = pdf_loader.load()

print(f"Loaded {len(pdf_documents)} pages from '{pdf_filename}'.")
pdf_loader = PyPDFLoader(pdf_filename)

# Load the documents from the PDF
pdf_documents = pdf_loader.load()

print(f"Loaded {len(pdf_documents)} pages from '{pdf_filename}'.")

# Display the content of each page (or the first page for brevity)
for i, doc in enumerate(pdf_documents):
    print(f"\n--- Content of Page {i+1} ---")
    print(doc.page_content[:500]) # Print first 500 characters for brevity
    print(f"Metadata: {doc.metadata}")
```
